Boolean Masking

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print(arr > 25)
```

```python
print(arr[arr>25])
```

```python
arr = np.array([10,20,30,40,50,60])

print(arr[(arr > 20) & (arr < 60)])
```

Fancy Indexing

```python
arr = np.array([100,200,300,400,500])

print(arr[[0,2,4]])
```

Flattening Array

```python
arr = np.array([
    [1,2,3],
    [4,5,6]
])

print(arr.flatten())
```

Transpose

```python
arr = np.array([
    [1,2,3],
    [4,5,6]
])

print(arr.T)
```

Axis Operations

```python
arr = np.array([
    [1,2,3],
    [4,5,6]
])
print(np.sum(arr, axis =0))
print(np.sum(arr, axis = 1))
```

Stacking Arrays

```python
a = np.array([1,2,3])
b = np.array([4,5,6])

np.vstack((a,b))
np.hstack((a,b))
```

```python
import numpy as np

arr = np.array([20,40,60,80,100])

normalized = (arr - arr.min()) / (arr.max() - arr.min())

print(normalized)
```
