1.Reverse a string without using slicing

```python
s = "Madhuri"

print(s[::-1])
```

2.Check Palindrome

```python
s=="Madhuri"
if s==s[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")    
```

3. Find Duplicate Elements

```python
nums=[1,2,3,4,5,2, 5, 5, 7, 0]
duplicate=[]
for i in nums:
    if nums.count(i)>1 and i not in duplicate:
        duplicate.append(i)
print(duplicate)
```

4. Find Frequency of Characters

```python
s = "python"

freq = {}

for char in s:
    freq[char] = freq.get(char,0) + 1

print(freq)
```

5. Find Second Largest Number

```python
nums = [10,20,5,40,30]

nums = list(set(nums))
nums.sort()

print(nums[-2])
```

6. Merge Two Dictionaries

```python
d1 = {"a":1}
d2 = {"b":2}

result = {**d1, **d2}

print(result)
```

7. Count Vowels

```python
s = "madhuri"

count = 0

for ch in s.lower():
    if ch in "aeiou":
        count += 1

print(count)
```

8. Remove Duplicates From List

```python
nums = [1,2,2,3,4,4,5]

result = list(set(nums))

print(result)
```

9. Find Largest Element

```python
nums = [10,50,20,100,5]

print(max(nums))
```
