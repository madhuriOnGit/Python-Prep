1. Reverse Words

```python
s = "I love Python"

print(" ".join(s.split()[::-1]))
```

2. Count Occurrence

```python
nums = [1,2,1,3,2,1]

freq = {}

for i in nums:
    freq[i] = freq.get(i,0)+1

print(freq)
```

```python
a =[1,2,3,4,5,6,]

b =[4,7, 9, 2, 0]

print(set(a) & set(b))
```

5. Check Anagram

6. Sum of Digits

```python
num = 12345
total =0
for i in str(num):
    total += int(i)
print(total)
```

7. Find Largest Word

```python
s = "I love learning Python"

words = s.split()

print(max(words,key=len))
```

8. Flatten List

```python
lst = [[1,2],[3,4],[5,6]]

result = []

for sub in lst:
    result.extend(sub)

print(result)
```

9. Count Upper and Lower Case

```python
s = "PyThOn"

upper = 0
lower = 0

for ch in s:
    if ch.isupper():
        upper += 1
    elif ch.islower():
        lower += 1

print(upper, lower)
```

10. Find Missing Number

```python
nums = [1,2,3,5]

n = 5

expected = n*(n+1)//2

actual = sum(nums)

print(expected-actual)
```
